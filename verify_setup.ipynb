# Vertify environment setup

```python
import sys
import importlib

required = {
    "numpy": "1.24",
    "pandas": "1.5",
    "sklearn": "1.2",
    "matplotlib": "3.6",
    "seaborn": "0.12",
    "Bio": "1.80",
}

print(f"Python version: {sys.version}\n")

all_ok = True
for module_name, min_version in required.items():
    try:
        mod = importlib.import_module(module_name)
        version = getattr(mod, "__version__", "unknown")
        print(f"✅  {module_name:12s}  version {version}")
    except ImportError:
        print(f"❌  {module_name:12s}  NOT FOUND")
        all_ok = False

print()
if all_ok:
    print("🎉 All packages found! You're ready to begin.")
else:
    print("⚠️  Some packages are missing. Re-run the pip install step.")
```

```python
import numpy as np
import pandas as pd
from Bio.Seq import Seq

sequence = "ATGCATGCATGCATGCATGCATGCATGC"
seq_obj = Seq(sequence)

mapping = {'A': 0, 'T': 1, 'G': 2, 'C': 3}
encoded = np.array([mapping[base] for base in sequence])

df = pd.DataFrame({
    'position': range(len(sequence)),
    'base': list(sequence),
    'encoded': encoded
})

print(f"Sequence length: {len(seq_obj)} bp")
print(f"GC content: {100 * (sequence.count('G') + sequence.count('C')) / len(sequence):.1f}%")
print(f"\nFirst 5 positions:\n{df.head()}")
print("\n✅ Core packages working correctly.")
```
